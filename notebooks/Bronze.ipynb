# Reading the CSV Files

```python
df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .csv("/Volumes/data_engineering_2026/bronze/source_system/Source_Crm/cust_info.csv")\

df.display()
```

```python
(
df.write
    .mode("overwrite")
    .format("delta")
    .saveAsTable("data_engineering_2026.bronze.crm_cust_info")
)
```
